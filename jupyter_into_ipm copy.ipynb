```python
base_url= "https://earthquake.usgs.gov/fdsnws/event/1/query"

full_url = base_url + "?format=geojson"

#### search - was there an earthquake at rally to restore sanity
full_url = (full_url
       + "&starttime=2010-10-30"
       + "&endtime=2010-10-31"
       + "&latitude=38.8"
       + "&longitude=-77"
       + "&maxradiuskm=1000"
        )


full_url
```
